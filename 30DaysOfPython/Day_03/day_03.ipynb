```python
import sympy as sym

x = sym.symbols('x')

expr = 2 * x - 2

slope1 = expr.coeff(x)

y_interceptor = expr.subs(x,0)

x_interceptor = sym.solve(sym.Eq(expr,0),x)[0]

print(f"slope {slope}")
print(f"x-interceptor {x_interceptor}")
print(f"y-interceptor {y_interceptor}")
```

```python
import math

point1 = (2,2)
point2 = (6,10)

slope2 = (point2[1] - point1[1])/(point2[0]-point1[0])

distance = math.sqrt((point2[0] - point1[0])**2 + (point2[1] - point1[1])**2)
print(f"The slope of two point is {slope}")
print(f"The Euclidean distance is {distance:.2f}")
```
